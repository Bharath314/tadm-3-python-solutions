### Exercise 1-5

**Q.** The knapsack problem is as follows: given a set of integers $S = \{s_1 , s_2 , . . . , s_n \}$,
and a target number $T$ , ﬁnd a subset of $S$ that adds up exactly to $T$. For example,
there exists a subset within $S = \{1, 2, 5, 9, 10\}$ that adds up to $T = 22$
but not $T = 23$.

Find counterexamples to each of the following algorithms for the knapsack problem.
That is, give an $S$ and $T$ where the algorithm does not ﬁnd a solution that
leaves the knapsack completely full, even though a full-knapsack solution exists.

**(a).** Put the elements of $S$ in the knapsack in left to right order if they ﬁt, that
is, the ﬁrst-ﬁt algorithm.

The first-fit algorithm can be implemented as follows:

````python
def first_fit(S: list[int], T: int):
    knapsack = []
    current_sum = 0
    for number in S:
        if number <= (T - current_sum):
            knapsack.append(number)
            current_sum += number
        if current_sum == T:
            break
    if current_sum == T:
        return knapsack
    return []
````

Take $S = \{1, 2, 3, 4\}$ and $T = 7$. The full-knapsack solution would be $\{1, 2, 4\}$.  However, upon running the first-fit algorithm on these values, we fail to find the solution.

````python
first_fit([1, 2, 3, 4], 7)
````

**(b).** Put the elements of S in the knapsack from smallest to largest,
that is, the best-ﬁt algorithm.

The best-fit algorithm can be implemented as follows:

````python
def best_fit(S: list[int], T: int):
    items = list(S)
    knapsack = []
    current_sum = 0
    while current_sum < T:
        gap = T - current_sum
        best_candidate = None
        for x in items:
            if x <= gap:
                if best_candidate is None or x > best_candidate:
                    best_candidate = x
        if best_candidate is not None:
            knapsack.append(best_candidate)
            current_sum += best_candidate
            items.remove(best_candidate)
        else:
            break
   
    if current_sum == T:
        return knapsack
    return []
````

Take $S = \{2, 3, 3, 4, 5\}$ and $T = 10$. The full-knapsack solution would be $\{2, 3, 5\}$.  However, upon running the best-fit algorithm on these values, we fail to find the solution.

````python
best_fit([2, 3, 3, 4, 5], 10)
````

**(c).** Put the elements of S in the knapsack from largest to smallest.

This algorithm can be implemented as follows:

````python
def first_fit_sorted(S: list[int], T: int):
    S = sorted(S, reverse=True)
    knapsack = []
    current_sum = 0
    for number in S:
        if number <= (T - current_sum):
            knapsack.append(number)
            current_sum += number
        if current_sum == T:
            break
    if current_sum == T:
        return knapsack
    return []
````

Take $S = \{2, 3, 3, 4, 5\}$ and $T = 10$. The full-knapsack solution would be $\{2, 3, 5\}$.  However, upon running the sorted first-fit algorithm on these values, we fail to find the solution.

````python
first_fit_sorted([2, 3, 3, 4, 5], 10)
````

**Q.** The set cover problem is as follows: given a set $S$ of subsets $S_1 , \dots, S_m$ of
the universal set $U = \{1, \dots, n\}$, find the smallest subset of subsets $T ⊆ S$ such
that $\cup_{t_i ∈ T}t_i = U$ . For example, consider the subsets $S_1 = \{1, 3, 5\}$, $S_2 = \{2, 4\}$, $S_3 = \{1, 4\}$, and $S_4 = \{2, 5\}$. The set cover of $\{1, \dots, 5\}$ would then be $S_1$ and $S_2$.

Find a counterexample for the following algorithm: Select the largest subset for
the cover, and then delete all its elements from the universal set. Repeat by
adding the subset containing the largest number of uncovered elements until all
are covered.

The algorithm can be implemented as follows:

````python
def greedy_set_cover(S: list[set[int]]):
    U = set().union(*S)
    cover = []
    while U:
        best_set = max(S, key=lambda s: len(s & U))
        cover.append(best_set)
        U -= best_set
        S.remove(best_set)
    return cover
````

Take $S = \{\{1,3,5\}, \{1,2\}, \{3,4\}, \{5,6\}\}$ and $T = 10$. The set cover would be $\{\{1,2\}, \{3,4\}, \{5,6\}\}$.  However, upon running the greedy set cover algorithm on this set, we fail to find the solution.

````python
greedy_set_cover([{1,3,5}, {1,2}, {3,4}, {5,6}])
````

The algorithm can be implemented as follows:

````python
def maximum_clique(G: tuple[set[str], list[tuple[str, str]]]):
    vertices = G[0]
    edges = G[1]
    graph_degrees = {vertex: 0 for vertex in vertices}
    neighbours: dict[str, set] = {vertex: set() for vertex in vertices}
    for pair in edges:
        graph_degrees[pair[0]] += 1
        graph_degrees[pair[1]] += 1
        neighbours[pair[0]].add(pair[1])
        neighbours[pair[1]].add(pair[0])

    vertices_ordered_by_degree = sorted(
        list(vertices), key=lambda vertex: graph_degrees[vertex], reverse=True
    )
    print(vertices_ordered_by_degree)
    print(neighbours)
    clique = set()
    for vertex in vertices_ordered_by_degree:
        clique_member = True
        for member in clique:
            if vertex not in neighbours[member]:
                clique_member = False
                break
        if clique_member:
            clique.add(vertex)
    return clique
````

### Exercise 1-8

**Q.** Prove the correctness of the following recursive algorithm to multiply two
natural numbers, for all integer constants $c ≥ 2$.

$$
\begin{aligned}
&\text{Multiply}(y, z) \\
&\quad \textit{if } z = 0 \textit{ then } \text{return}(0) \textit{ else} \\
&\quad \text{return}(\text{Multiply}(cy, \lfloor z/c \rfloor) + y \cdot (z \bmod c))
\end{aligned}
$$

**Base Case:**

Taking $z=0$, $\text{Multiply}(y,z) = 0$, which is correct as $y \times 0 = 0$

Thus, $\text{Multiply}$ is proven to work on the basis case


**Inductive step:**

To prove the correctness of the function for $y = y^*$ and $z = z^*$, we assume that the function works correctly for the next recursive call where $y = cy^*$ and $z = \lfloor z^*/c \rfloor$, *i.e.*, $\text{Multiply}(cy^*, \lfloor z^*/c \rfloor) = cy^* \times \lfloor z^*/c \rfloor$.

$$
\begin{align*}
\text{Multiply}(y^*, z^*) &= \text{Multiply}(cy^*, \lfloor z^*/c \rfloor) + y^* \cdot (z^* \bmod c) \\
&= cy^* \cdot \lfloor z^*/c \rfloor + y^* \cdot (z^* \bmod c) \\
&= cy^* \cdot \frac{z^* - z^* \bmod c}{c} + y^* \cdot (z^* \bmod c) \\
&= y^* z^* - y^*(z^* \bmod c) + y^* \cdot (z^* \bmod c) \\
&= y^* z^*
\end{align*}
$$

Thus, by induction, we have proved the correctness of $\text{Multiply}(y,z)$

### Exercise 1-9

**Q.** Prove the correctness of the following algorithm for evaluating a polynomial $a_n x^n + a_{n−1} x^{n−1} + \dots + a_1 x + a_0$.
$$
\begin{align*}
&\text{Horner}(a, x) \\
&\quad p = a_n \\
&\quad \text{for } i \text{ from } n − 1 \text{ to } 0 \\
&\quad\quad p = p · x + a_i \\
&\quad \text{return } p
\end{align*}
$$

In every iteration of the loop, 

**Base case:**

At $i = n - 1$, we get $p = a_n x + a_{n-1}$, which can be written as $p = a_n x^{n - (n - 1)} + a_{n - 1} x^{(n - 1) - (n - 1)}$


**Inductive Step:**

To prove correctness of the function for $i = 0$, we assume that the function works correctly for the step, where $i = 1$, _i.e.,_,
$p = a_n x^{n - 1} + a_{n - 1} x^{(n - 1) - 1} + \dots + a_1 x^{1 - 1}$, which can be written as $p = a_n x^{n - 1} + a_{n - 1} x^{n - 2} + \dots + a_1$.

At $i = 0$:

$$
\begin{align*}
p &= p \cdot x + a_0 \\
p &= (a_n x^{n - 1} + a_{n - 1} x^{n - 2} + \dots + a_1) \cdot x + a_0 \\
p &= a_n x^n + a_{n - 1} x^{n - 1} + \dots + a_1 x + a_0 \\
p &= a_n x^{n - 0} + a_{n - 1} x^{(n - 1) - 0} + \dots + a_1 x^{1 - 0} + a_0 x^{0 - 0}
\end{align*}
$$


Thus, by induction, we have proved the correctness of Horner's algorithm.

### Exercise 1-10

**Q.** Prove the correctness of the following sorting algorithm.
$$
\begin{align*}
&\text{Bubblesort } (A) \\
&\quad\text{for } i \text{ from } n \text{ to } 1 \\
&\quad\quad\text{for } j \text{ from } 1 \text{ to } i − 1 \\
&\quad\quad\quad\text{if } (A[j] > A[j + 1]) \\
&\quad\quad\quad\quad\text{swap the values of } A[j] \text{ and } A[j + 1] \\
\end{align*}
$$

**1. Inner Loop Induction**

**Claim:** After the inner loop ($j$-loop) runs for a fixed $i$, $A[i]$ contains the largest element in $A[1 \dots i]$.

**Base Case:** $j = 1$. We compare $A[1]$ and $A[2]$. The larger of the two is moved to $A[2]$. Thus, $A[2]$ is the maximum of $A[1 \dots 2]$.

**Inductive Step:** Assume that after $k$ iterations, $A[k+1]$ is the maximum of $A[1 \dots k+1]$.

**Next Step:** In iteration $k+1$, we compare $A[k+1]$ and $A[k+2]$. Since $A[k+1]$ was already the max of the previous elements, the winner of this comparison must be the maximum of $A[1 \dots k+2]$.

By induction, when $j$ reaches $i-1$, the maximum element of the subarray $A[1 \dots i]$ has been "bubbled" up to position $A[i]$.

**2. Outer Loop Induction**

**Claim:** After $k$ iterations of the outer loop ($i$-loop), the last $k$ elements of the array are the $k$ largest elements in sorted order.

**Base Case:** $k = 1$ (which corresponds to $i = n$). The inner loop runs once and, as proved above, the largest element of $A[1 \dots n]$ is placed at $A[n]$. The claim holds.

**Inductive Step:** Assume that after $k$ iterations (where $i = n, n-1, \dots, n-k+1$), the subarray $A[n-k+1 \dots n]$ contains the $k$ largest elements in sorted order.

**Next Step:** In the next iteration ($i = n-k$), the inner loop runs from $1$ to $i-1$. By our inner induction, it finds the maximum of the remaining unsorted elements $A[1 \dots n-k]$ and places it at $A[n-k]$.

Since this new element is the largest of the remaining set, and all elements to its right are already larger and sorted, the subarray $A[n-k \dots n]$ now contains the $k+1$ largest elements in sorted order.

**3. Final Termination**

When the outer loop finishes (when $i=1$), the induction tells us that $A[2 \dots n]$ contains the $n-1$ largest elements in sorted order. By necessity, the single element left at $A[1]$ must be the smallest, meaning the entire array is sorted.

### Exercise 1-11

**Q.** The greatest common divisor of positive integers $x$ and $y$ is the largest integer
$d$ such that d divides $x$ and d divides $y$. Euclid’s algorithm to compute $\text{gcd}(x, y)$
where $x > y$ reduces the task to a smaller problem:

$$\text{gcd}(x, y) = \text{gcd}(y, x \bmod y)$$

Prove that Euclid’s algorithm is correct.

**Base Case** ($y = 1$)

We test the smallest possible positive integer value for $y$. If $y = 1$, then for any $x > 1$:

$$x = x \cdot 1 + 0 \implies r = x \bmod 1 = 0$$

According to the algorithm's base terminal condition, when the remainder is $0$, the current divisor is the answer:

$$\gcd(x, 1) = \gcd(1, 0) = 1$$

Since $1$ is indeed the greatest common divisor of any integer $x$ and $1$, $P(1)$ is true.

**Inductive Hypothesis**

Assume that the algorithm works perfectly for all values of $y$ up to some integer $k$.That is, we assume $P(1), P(2), \dots, P(k)$ are all true. This means for any pair where the second element is strictly less than or equal to $k$, Euclid's algorithm yields the correct greatest common divisor.

**Inductive Step**

We now need to prove that $P(k+1)$ is true. Consider a pair $(x, k+1)$ where $x > k+1$. Applying the division step of Euclid's algorithm:

$$\gcd(x, k+1) = \gcd(k+1, x \bmod (k+1))$$

Let $r = x \bmod (k+1)$. By the properties of modular arithmetic:

$$0 \leq r < k+1$$

We have two possibilities for $r$: 
1. If $r = 0$: The algorithm terminates immediately and returns $k+1$. Since $k+1$ divides both itself and $x$ (because the remainder is $0$), $k+1$ is definitively the correct $\gcd$.
2. If $r > 0$: The algorithm reduces the problem to evaluating $\gcd(k+1, r)$.Because $0 < r < k+1$, the new second argument ($r$) falls strictly within our inductive hypothesis range ($1 \le r \le k$). 

Therefore, by our inductive hypothesis, the algorithm is guaranteed to find the correct value for $\gcd(k+1, r)$.Since $\gcd(x, k+1) = \gcd(k+1, r)$ and the algorithm evaluates the latter flawlessly, it must also evaluate $\gcd(x, k+1)$ correctly.Thus, $P(k+1)$ is true.

By the principle of strong induction, since the base case holds and $P(1), \dots, P(k) \implies P(k+1)$, the proposition $P(y)$ is true for all positive integers $y$. Euclid's algorithm is correct.
